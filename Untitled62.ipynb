```python
# 2. Clean/Remap labels to classes: worker, helmet, vest, boot
import os, shutil, yaml, glob, re
from pathlib import Path

SRC = Path("/content/dataset_ppe")   # where we extracted
DST = Path("/content/dataset_ppe_clean")
DST.mkdir(exist_ok=True)

# Desired classes
NEW_NAMES = ["worker", "helmet", "vest", "boot"]
new_name_to_id = {n:i for i,n in enumerate(NEW_NAMES)}

# Helper: find data.yaml if present
def find_yaml(root):
    cand = list(root.glob("**/data*.yaml")) + list(root.glob("**/data*.yml"))
    return cand[0] if cand else None

data_yaml = find_yaml(SRC)
orig_names = None
train_images = None
val_images = None

if data_yaml:
    print("Found data yaml:", data_yaml)
    with open(data_yaml, 'r') as f:
        d = yaml.safe_load(f)
    # d may contain 'names' as list/dict and 'train','val' paths
    if isinstance(d.get("names"), dict):
        # convert dict to list by index
        names = [d["names"][i] for i in sorted(map(int, d["names"].keys()))]
    else:
        names = d.get("names")
    orig_names = names
    train_images = d.get("train")
    val_images = d.get("val")
    print("Original dataset names:", orig_names)
else:
    print("No data.yaml found. Will try to auto-detect folder structure.")

# Find images and labels folders (common patterns)
def find_images_labels(root):
    # common layout: images/train, images/val, labels/train, labels/val
    imgs = {}
    lbls = {}
    for split in ["train","val","test"]:
        for base in ["images","img","JPEGImages","imgs"]:
            img_dir = root / base / split
            lbl_dir = root / "labels" / split
            if img_dir.exists() and lbl_dir.exists():
                imgs[split] = img_dir
                lbls[split] = lbl_dir
    # fallback: if top-level images and labels exist
    if not imgs:
        for base in ["images","img","JPEGImages","imgs"]:
            if (root / base).exists() and (root / "labels").exists():
                imgs["all"] = root / base
                lbls["all"] = root / "labels"
    # fallback: just use root images
    if not imgs:
        for p in root.rglob("images"):
            lblp = p.parent / "labels"
            if lblp.exists():
                imgs["all"] = p
                lbls["all"] = lblp
    return imgs, lbls

imgs, lbls = find_images_labels(SRC)
print("Detected image/label splits:", imgs.keys())

# Utility to read a label line robustly (keep first 5 numbers if rotated boxes exist)
def parse_label_line(line):
    parts = re.split(r'\s+', line.strip())
    if not parts or parts == ['']:
        return None
    cls = int(parts[0])
    nums = parts[1:]
    # keep first 4 numbers if available (x,y,w,h) and ignore extras (angle) if present
    if len(nums) >= 4:
        return cls, nums[:4]
    else:
        return None

# If orig_names available -> build mapping from original class name -> new class id OR 'remove'
remap = {}
if orig_names:
    for idx, n in enumerate(orig_names):
        name = n.lower().strip()
        if name in NEW_NAMES:
            remap[idx] = new_name_to_id[name]
        elif name.startswith("no ") or name.startswith("no-") or "no" in name.split():
            # anything like "no helmet" -> remove
            remap[idx] = None
        else:
            # unknown labels -> remove (safer)
            remap[idx] = None
    print("Remap table (orig idx -> new idx or None=drop):", remap)
else:
    # If we don't have YAML, try to detect textual class names in possible .names/.txt files
    # Fallback: assume label files already use indices where:
    # 0: worker, 1: helmet, 2: vest, 3: boot  (this is a guess)
    print("No original names found: we will attempt to keep only classes with index 0-3 (you may need to edit this mapping).")
    remap = {0:0, 1:1, 2:2, 3:3}  # best effort

# Now build the cleaned dataset folder structure
# Strategy: copy images across preserving train/val if present; rewrite label txt files with new class ids or drop lines.
def copy_and_remap(img_dir, label_dir, out_images_dir, out_labels_dir):
    out_images_dir.mkdir(parents=True, exist_ok=True)
    out_labels_dir.mkdir(parents=True, exist_ok=True)
    img_files = list(img_dir.glob("**/*"))
    # consider typical image extensions
    valid_exts = {'.jpg','.jpeg','.png','.bmp','.tif','.tiff'}
    for imgf in img_files:
        if imgf.suffix.lower() in valid_exts:
            rel = imgf.relative_to(img_dir)
            dest_img = out_images_dir / rel
            dest_img.parent.mkdir(parents=True, exist_ok=True)
            shutil.copy2(imgf, dest_img)
            # corresponding label file
            lbl = label_dir / rel.with_suffix('.txt')
            if lbl.exists():
                with open(lbl, 'r') as rf:
                    lines = rf.readlines()
                new_lines = []
                for L in lines:
                    parsed = parse_label_line(L)
                    if parsed is None:
                        continue
                    clsidx, coords = parsed
                    newcls = remap.get(clsidx, None)
                    if newcls is None:
                        continue
                    new_lines.append(f"{newcls} {' '.join(map(str, coords))}\n")
                if new_lines:
                    with open(out_labels_dir / rel.with_suffix('.txt'), 'w') as wf:
                        wf.writelines(new_lines)
                else:
                    # no labels after remap -> write empty file (YOLO expects either present or missing)
                    open(out_labels_dir / rel.with_suffix('.txt'), 'w').close()
            else:
                # no label -> empty label file
                open(out_labels_dir / rel.with_suffix('.txt'), 'w').close()

# Case: detected train/val
if imgs:
    for split, img_dir in imgs.items():
        lbl_dir = lbls.get(split)
        if not lbl_dir:
            print("Warning: labels dir for", split, "not found")
            continue
        out_img_dir = DST / "images" / split
        out_lbl_dir = DST / "labels" / split
        print("Processing", split, "images from", img_dir, "labels from", lbl_dir)
        copy_and_remap(img_dir, lbl_dir, out_img_dir, out_lbl_dir)
else:
    print("No images/labels found — check dataset structure under", SRC)

# Write new data.yaml for ultralytics
data = {
    "train": str((DST / "images" / "train").resolve()) if (DST / "images" / "train").exists() else str((DST / "images").resolve()),
    "val": str((DST / "images" / "val").resolve()) if (DST / "images" / "val").exists() else str((DST / "images").resolve()),
    "names": NEW_NAMES
}
with open(DST / "data.yaml", 'w') as f:
    yaml.safe_dump(data, f)
print("Wrote cleaned dataset to", DST)
print("New data.yaml:")
print(open(DST / "data.yaml").read())
```

```python
# 4. Image inference + Safe/Unsafe per worker
import cv2, numpy as np
from ultralytics import YOLO
import torch
from pathlib import Path

WEIGHTS = "//content/yolov8n.pt"
IMAGE_PATH = "/content/Personal-Protective-Equipment-678x381.jpg"
# Class names must match the order we wrote in data.yaml
CLASS_NAMES = ["worker", "helmet", "vest", "boot"]
IDX = {n:i for i,n in enumerate(CLASS_NAMES)}

model = YOLO(WEIGHTS)

# Run inference (single image)
res = model.predict(source=IMAGE_PATH, imgsz=640, conf=0.25, iou=0.5)[0]

# Extract boxes, classes
boxes = res.boxes  # Boxes object
xyxy = boxes.xyxy.cpu().numpy()        # (N,4) x1,y1,x2,y2
cls = boxes.cls.cpu().numpy().astype(int)  # (N,)
conf = boxes.conf.cpu().numpy()

# helper func: center point
def center_of_box(b):
    x1,y1,x2,y2 = b
    return ((x1+x2)/2, (y1+y2)/2)

# IoU helper
def iou_xyxy(a,b):
    x1 = max(a[0], b[0]); y1 = max(a[1], b[1])
    x2 = min(a[2], b[2]); y2 = min(a[3], b[3])
    inter_w = max(0, x2-x1); inter_h = max(0, y2-y1)
    inter = inter_w * inter_h
    area_a = (a[2]-a[0])*(a[3]-a[1])
    area_b = (b[2]-b[0])*(b[3]-b[1])
    union = area_a + area_b - inter
    return inter/union if union>0 else 0.0

# Gather detections by class
detections = []
for b,cf,c in zip(xyxy, conf, cls):
    detections.append({"box": b, "cls": int(c), "conf": float(cf)})

workers = [d for d in detections if d["cls"] == IDX["worker"]]
ppe = [d for d in detections if d["cls"] in (IDX["helmet"], IDX["vest"], IDX["boot"])]

# For each worker, check presence of each PPE
results_per_worker = []
for w in workers:
    wbox = w["box"]
    found = {"helmet": False, "vest": False, "boot": False}
    for p in ppe:
        pname = CLASS_NAMES[p["cls"]]
        # criterion: PPE center inside worker box OR IoU > 0.05
        cx,cy = center_of_box(p["box"])
        if (cx >= wbox[0] and cx <= wbox[2] and cy >= wbox[1] and cy <= wbox[3]) or iou_xyxy(wbox, p["box"]) > 0.05:
            found[pname] = True
    safe = all(found.values())
    results_per_worker.append({"worker_box": wbox, "found": found, "safe": safe})

# Draw visualization
img = cv2.imread(IMAGE_PATH)
for r in results_per_worker:
    x1,y1,x2,y2 = map(int, r["worker_box"])
    label = "SAFE" if r["safe"] else "UNSAFE"
    color = (0,255,0) if r["safe"] else (0,0,255)
    cv2.rectangle(img, (x1,y1), (x2,y2), color, 2)
    cv2.putText(img, label, (x1, y1-10), cv2.FONT_HERSHEY_SIMPLEX, 1.0, color, 2)
    # draw small text of found PPE
    text = ",".join([k for k,v in r["found"].items() if v])
    cv2.putText(img, text, (x1, y2+20), cv2.FONT_HERSHEY_SIMPLEX, 0.7, (255,255,255), 2)

# Save and display (Colab)
OUT = "/content/ppe_result2.jpg"
cv2.imwrite(OUT, img)
print("Wrote", OUT)
from IPython.display import Image, display
display(Image(OUT))
```

```python
# 5A. Per-frame video annotation (no tracking)
import cv2, numpy as np
from ultralytics import YOLO

WEIGHTS = "/content/yolov8n.pt"
VIDEO_IN = "/content/WhatsApp Video 2025-09-29 at 12.43.44.mp4"   # change
VIDEO_OUT = "/content/ppe_annotated_output.avi"

model = YOLO(WEIGHTS)
cap = cv2.VideoCapture(VIDEO_IN)
fourcc = cv2.VideoWriter_fourcc(*'XVID')
fps = cap.get(cv2.CAP_PROP_FPS) or 20.0
w = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
h = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
out = cv2.VideoWriter(VIDEO_OUT, fourcc, fps, (w,h))

frame_no = 0
while True:
    ret, frame = cap.read()
    if not ret:
        break
    frame_no += 1
    # run inference per frame (stream=True yields faster memory usage)
    # convert frame to RGB, save temp file or use model.predict on numpy by passing frame directly (Ultralytics supports np arrays)
    res = model.predict(source=frame, conf=0.25, imgsz=640, verbose=False)[0]
    boxes = res.boxes
    xyxy = boxes.xyxy.cpu().numpy()
    cls = boxes.cls.cpu().numpy().astype(int)
    conf = boxes.conf.cpu().numpy()

    # build same safe/unsafe logic as above
    detections = [{"box": b, "cls":int(c), "conf":float(cf)} for b,cf,c in zip(xyxy, conf, cls)]
    workers = [d for d in detections if d["cls"] == IDX["worker"]]
    ppe = [d for d in detections if d["cls"] in (IDX["helmet"], IDX["vest"], IDX["boot"])]

    annotated = frame.copy()
    for w in workers:
        wbox = w["box"]
        found = {"helmet": False, "vest": False, "boot": False}
        for p in ppe:
            pname = CLASS_NAMES[p["cls"]]
            cx,cy = ( (p["box"][0]+p["box"][2])/2, (p["box"][1]+p["box"][3])/2 )
            if (cx >= wbox[0] and cx <= wbox[2] and cy >= wbox[1] and cy <= wbox[3]) or iou_xyxy(wbox, p["box"]) > 0.05:
                found[pname] = True
        safe = all(found.values())
        x1,y1,x2,y2 = map(int, wbox)
        color = (0,255,0) if safe else (0,0,255)
        label = "SAFE" if safe else "UNSAFE"
        cv2.rectangle(annotated, (x1,y1),(x2,y2), color, 2)
        cv2.putText(annotated, label, (x1, max(10,y1-8)), cv2.FONT_HERSHEY_SIMPLEX, 0.8, color, 2)
    out.write(annotated)

cap.release()
out.release()
print("Wrote annotated video to", VIDEO_OUT)
```
