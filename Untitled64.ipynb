```python
from ultralytics import YOLO
import cv2
import matplotlib.pyplot as plt
import numpy as np

model = YOLO("/content/runs/detect/ppe_detector/weights/best.pt")

# Classes in your dataset
classes = ["helmet", "no helmet", "vest", "no vest", "boot", "no boot", "worker"]

def is_overlapping(worker_box, ppe_box, threshold=0.1):
    x1 = max(worker_box[0], ppe_box[0])
    y1 = max(worker_box[1], ppe_box[1])
    x2 = min(worker_box[2], ppe_box[2])
    y2 = min(worker_box[3], ppe_box[3])
    inter_area = max(0, x2 - x1) * max(0, y2 - y1)
    ppe_area = (ppe_box[2] - ppe_box[0]) * (ppe_box[3] - ppe_box[1])
    return (inter_area / (ppe_area + 1e-6)) > threshold

def classify_workers(image_path, model, classes):
    results = model.predict(image_path, imgsz=640, conf=0.25)
    result = results[0]
    img = cv2.imread(image_path)

    worker_boxes = []
    ppe_boxes = []

    for box, cls_id in zip(result.boxes.xyxy, result.boxes.cls):
        cls_id = int(cls_id)
        label = classes[cls_id]
        xyxy = box.cpu().numpy().astype(int)

        if label == "worker":
            worker_boxes.append((xyxy, []))
        elif label in ["helmet", "vest", "boot"]:
            ppe_boxes.append((label, xyxy))
        # ignore "no helmet", "no vest", "no boot"

    # Assign PPE to workers
    for i, (wbox, w_ppe) in enumerate(worker_boxes):
        for label, pbox in ppe_boxes:
            if is_overlapping(wbox, pbox):
                w_ppe.append(label)
        worker_boxes[i] = (wbox, w_ppe)

    # Draw SAFE/UNSAFE
    for idx, (wbox, w_ppe) in enumerate(worker_boxes, start=1):
        x1, y1, x2, y2 = wbox
        has_helmet = "helmet" in w_ppe
        has_vest = "vest" in w_ppe
        has_boot = "boot" in w_ppe

        status = "SAFE" if has_helmet and has_vest and has_boot else "UNSAFE"
        color = (0, 255, 0) if status == "SAFE" else (0, 0, 255)

        cv2.rectangle(img, (x1, y1), (x2, y2), color, 2)
        cv2.putText(img, f"{status} #{idx}", (x1, y1 - 10),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.8, color, 2)

    return img

# Test
image_path = "/content/images.png"
output_img = classify_workers(image_path, model, classes)
plt.imshow(cv2.cvtColor(output_img, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```
