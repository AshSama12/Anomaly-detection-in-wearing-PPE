```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow  # if you're in Colab

def check_and_draw_worker_safety(results, image_path=None, save_path=None, conf_threshold=0.0):
    """
    - results: ultralytics model.predict(...) output
    - image_path: optional, if provided the function will draw SAFE/UNSAFE on this image
    - save_path: if provided, save the annotated image
    - conf_threshold: minimum box confidence to consider (0.0 to accept all)
    Returns: (total_safe, total_unsafe)
    """
    img = cv2.imread(image_path) if image_path else None
    total_safe = 0
    total_unsafe = 0

    for r in results:
        boxes = r.boxes
        names = r.names

        workers = []
        ppe_dets = []  # list of dicts: {'cls':..., 'xyxy':..., 'conf':...}

        for box in boxes:
            cls_name = names[int(box.cls)]
            xyxy = box.xyxy[0].cpu().numpy().astype(int)
            # get confidence if available
            try:
                conf = float(box.conf[0])
            except Exception:
                conf = 1.0

            if conf < conf_threshold:
                continue

            if cls_name == "worker":
                workers.append({'xyxy': xyxy, 'conf': conf})
            elif "helmet" in cls_name or "vest" in cls_name or "boot" in cls_name:
                ppe_dets.append({'cls': cls_name, 'xyxy': xyxy, 'conf': conf})

        def center_inside(p_xyxy, w_xyxy):
            px1, py1, px2, py2 = p_xyxy
            cx = (px1 + px2) / 2.0
            cy = (py1 + py2) / 2.0
            wx1, wy1, wx2, wy2 = w_xyxy
            return (cx >= wx1) and (cx <= wx2) and (cy >= wy1) and (cy <= wy2)

        # for each worker determine presence for each PPE type
        for w in workers:
            wxy = w['xyxy']

            def best_conf_for_type(base_name):
                # returns (best_pos_conf, best_neg_conf)
                pos_name = base_name
                neg_name = "no " + base_name
                best_pos = 0.0
                best_neg = 0.0
                for d in ppe_dets:
                    if not center_inside(d['xyxy'], wxy):
                        continue
                    # exact match of class string; if your labels use different tokens adjust here
                    if d['cls'] == pos_name:
                        best_pos = max(best_pos, d['conf'])
                    elif d['cls'] == neg_name:
                        best_neg = max(best_neg, d['conf'])
                return best_pos, best_neg

            hp, hn = best_conf_for_type('helmet')
            vp, vn = best_conf_for_type('vest')
            bp, bn = best_conf_for_type('boot')

            # decide presence: prefer positive if its confidence > negative, else negative wins
            has_helmet = (hp > hn) and (hp > 0)
            has_vest   = (vp > vn) and (vp > 0)
            has_boot   = (bp > bn) and (bp > 0)

            safe = has_helmet and has_vest and has_boot

            if safe:
                total_safe += 1
            else:
                total_unsafe += 1

            # draw on image if available
            if img is not None:
                x1, y1, x2, y2 = wxy
                color = (0, 255, 0) if safe else (0, 0, 255)
                cv2.rectangle(img, (x1, y1), (x2, y2), color, 2)
                label = "SAFE" if safe else "UNSAFE"
                cv2.putText(img, label, (x1, max(y1-8, 10)), cv2.FONT_HERSHEY_SIMPLEX, 0.9, color, 2)

    if save_path and img is not None:
        cv2.imwrite(save_path, img)
        print(f"Annotated image saved to: {save_path}")

    # show image in Colab (optional)
    if img is not None:
        try:
            cv2_imshow(img)
        except Exception:
            pass

    print(f"Total Safe: {total_safe}, Total Unsafe: {total_unsafe}")
    return total_safe, total_unsafe
```

```python
# run inference on an image
image_path = "/content/dataset/test/images/WhatsApp-Image-2024-04-21-at-2_37_06-PM-1-_jpeg.rf.cdb8d7338d27df44be3919f5bb40a48a.jpg"
results = model.predict(source=image_path, conf=0.25)

# check worker safety and draw result
check_and_draw_safety(results, image_path, save_path="worker_safety.jpg")

# show the output image
from google.colab.patches import cv2_imshow
output_img = cv2.imread("worker_safety.jpg")
cv2_imshow(output_img)
```
